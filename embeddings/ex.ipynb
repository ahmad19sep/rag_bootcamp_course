```python
from langchain_huggingface import HuggingFaceEmbeddings
import numpy as np
```

```python

documents = [
    "Employees receive 20 paid vacation days every year.",
    "Python is widely used in machine learning.",
    "The company uses PostgreSQL as its main database.",
    "Workers can work remotely two days per week."
]

query = "How many annual leave days do staff get?"

```

```python
huggingface_embeddings=HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2",
)
```

```python
def create_doc_embeddings(documents):
    return huggingface_embeddings.embed_documents(documents)

```

```python
def create_query_embeddings(query):
    return huggingface_embeddings.embed_query(query)
```

```python
def calculate_cosine_similarity(query_embeddings,doc_embeddings):
    dot_product=np.dot(query_embeddings,doc_embeddings)
    query_norm=np.linalg.norm(query_embeddings)
    doc_norm=np.linalg.norm(doc_embeddings)
    return dot_product/(query_norm*doc_norm)



```

```python
def semantic_search(query,documents,top_k=2):
    query_embeddings=create_query_embeddings(query)
    doc_embeddings=create_doc_embeddings(documents)

    similarities=[]
    for i , doc in enumerate(doc_embeddings):
        similarity=calculate_cosine_similarity(query_embeddings,doc)
        print(f"{query} vs {documents[i]} : {similarity}")
        similarities.append((similarity,documents[i]))

    similarities.sort(reverse=True)
    return similarities[:top_k]





```

```python
results=semantic_search(query,documents) 
print(query)
for similarity,document in results:
    print(f"Similarity: {similarity:.2f} - Document: {document}")





```
