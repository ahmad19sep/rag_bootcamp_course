```python
from langchain_community.document_loaders import DirectoryLoader,TextLoader

dic_loader=DirectoryLoader(
 "D:/rag-bootcamp/text_files",
 glob="**/*.txt",
 loader_cls=TextLoader,
 loader_kwargs={'encoding': 'utf-8'},
 show_progress=True

)
```

```python
documents=dic_loader.load()

print(f"Loaded {len(documents)} documents")
for i , doc in enumerate[Any](documents):
    print(f"\n Document {i+1}: ")
    print(f"Source: {doc.metadata['source']}")
    print(f"Length: {len(doc.page_content)} characters")
```

```python
text=documents[1].page_content
text
```

```python
print("CHARACTER TEXTSPLITTER")
char_splitter_space=CharacterTextSplitter(
 separator=" ",
 chunk_size=200,
 chunk_overlap=20,
 length_function=len

)
```

```python
char_chunks=char_splitter_space.split_text(text)
print(f"Created {len(char_chunks)}")
```

```python
print("Line based splitter")
char_splitter_line=CharacterTextSplitter(
    separator="\n",
    chunk_size=200,
    chunk_overlap=20,
    length_function=len

)
```

```python
char_chunks=char_splitter_line.split_text(text)
print(f"created {len(char_chunks)}")
```

```python
print("RecursiveCharacterTextSplitter")

recursive_splitter=RecursiveCharacterTextSplitter(
 separators=[" "],
 chunk_size=200,
 chunk_overlap=20,
 length_function=len

)
recursive_chunks=recursive_splitter.split_text(text)
print(f"Created {len(recursive_chunks)}")


```

```python
print(recursive_chunks[0])
print("---------------")
print(recursive_chunks[1])
print("---------------")
print(recursive_chunks[2])
print("---------------")
```

```python
print("Token text splitter")
token_splitter=TokenTextSplitter(
 chunk_size=50,
 chunk_overlap=10
 

)

token_chunks=token_splitter.split_text(text)
print(f"create {len(token_chunks)}")


```
